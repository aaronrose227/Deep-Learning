```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

class GridWorld:
    def __init__(self, grid_size=(7, 10), start_state=(3, 0), goal_state=(3, 7), wind = [0, 0, 0, 1, 1, 1, 2, 2, 1, 0]):
        self.grid_size = grid_size
        self.start_state = start_state
        self.goal_state = goal_state
        self.wind = wind
        self.reset()

    def reset(self):
        """Resets the environment to the start state."""
        self.agent_pos = self.start_state
        return self.agent_pos

    def step(self, action):
        """Takes a step in the environment based on the given action."""
        x, y = self.agent_pos
        x = max(0, x - self.wind[y])
        if action == 0:  # Up
            x = max(0, x - 1)
        elif action == 1:  # Right
            y = min(self.grid_size[1] - 1, y + 1)
        elif action == 2:  # Down
            x = min(self.grid_size[0] - 1, x + 1)
        elif action == 3:  # Left
            y = max(0, y - 1)
        self.agent_pos = (x, y)
        reward = -1
        done = self.agent_pos == self.goal_state
        return self.agent_pos, reward, done

    def render(self, q_table):
        """Renders the grid world and overlays the optimal state values and the resulting policy."""
        grid = np.zeros(self.grid_size)
        grid[self.goal_state] = 2  # Mark goal
        grid[self.start_state] = -1  # Mark agent position
        plt.imshow(grid, cmap="coolwarm", interpolation="nearest")

        if q_table is not None:
            for i in range(self.grid_size[0]):
                for j in range(self.grid_size[1]):
                    if (i, j) != self.goal_state:
                        action = np.argmax(q_table[(i, j)])
                        arrow = {0: "↑", 1: "→", 2: "↓", 3: "←"}[action]
                        state_value = str(round(max(q_table[(i, j)]), 1))
                        plt.text(j, i, state_value + " " + arrow, ha="center", va="center", color="black")

        plt.xticks([]), plt.yticks([])
        plt.show()


# Code for an RL agent that interacts with the environment to learn the optimal policy
class RLAgent:
    def __init__(self, env, learning_rate=0.1, discount_factor=1, epsilon=0.1):
        self.env = env
        self.q_table = np.zeros((*env.grid_size, 4))  # Q-table for (state, action)
        self.learning_rate = learning_rate
        self.discount_factor = discount_factor
        self.epsilon = epsilon

    def choose_action(self, state):
        """Epsilon-greedy action selection."""
        if np.random.uniform() < self.epsilon:
            return np.random.choice(range(4))
        else:
            return np.argmax(self.q_table[state])

    def train_with_Q_learning(self, num_episodes=500):
        """Implement Q-Learning to learn the optimal policy"""
        for episode in tqdm(range(num_episodes)):
            state = self.env.reset()
            done = False
            while not done:
                pass

    def train_with_SARSA(self, num_episodes=500):
        """Implement SARSA to learn the optimal policy"""
        for episode in tqdm(range(num_episodes)):
            pass

    def train_with_MC(self, num_episodes=500):
        """Optional: Implement On-Policy First-Visit Monte-Carlo Control (see page 101 in Chapter 5.4 in Sutton and Barto) to learn the optimal policy. """
        for episode in tqdm(range(num_episodes)):
            pass
        
    def get_policy(self):
        """Extracts the optimal policy from the Q-table."""
        policy = np.argmax(self.q_table, axis=2)
        return policy


# Visualization
def visualize_policy(agent):
    policy = agent.get_policy()
    print("Optimal policy:")
    agent.env.render(agent.q_table)

# Example usage
env = GridWorld()   
agent = RLAgent(env)
#agent.train_with_Q_learning(num_episodes=1000)
agent.train_with_SARSA(num_episodes=1000)
#agent.train_with_MC(num_episodes=1000)
visualize_policy(agent)
```
